```python
path =r'/Volumes/batch28_catalog/test_b28_schema/bronze/customers-2000000.csv'
df = spark.read.format('csv').option('header','true').option('inferSchema','true').load(path)
df.display()
```

```python
from pyspark.sql.functions import *
df1= df.withColumn("full_name",concat(col('First Name'),lit(' '),(col('Last Name'))))
df1.display()
```

```python
# concat :- this is function used to merge two or more string colums
# concat_ws :- this is also merge two string columns but with specific separators 
from pyspark.sql.functions import *
df1= df.withColumn("full_name",concat_ws("-",col('First Name'),(col('Last Name'))))
df1.display()
```

```python
df2=df1.withColumn("user_name",(split(col('Email'),'@').getItem(0)))\
        .withColumn("domain",split(col('Email'),'@').getItem(1))
df2.display()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

# Define schema with three columns
data = [
    ("  John   Doe  ", 25),
    ("  Alice   Smith ", 30),
    ("  Bob   Lee  ", 35),
    ("  Charlie   Brown ", 28),
    ("  Diana   Prince ", 22)
]

schema = StructType([
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True)
])

df = spark.createDataFrame(data, schema)
df.display()


```

```python
# trim :- remove lefy and right side extra space in string
df1=df.select(trim(col('name')))
df1.display()
```

```python
# ltrim :- remove onlt left  side extra space in string
df1=df.select(ltrim(col('name')))
df1.display()
```

```python
# rtrim :- remove onlt right  side extra space in string
df1=df.select(rtrim(col('name')))
df1.display()
```

```python
# apply sql query on dataframe 
# spark.sql - It provide us facility that on dtaframe itself we can apply sql query
df.createOrReplaceTempView("customer")

df1=spark.sql("select Country,count(*) from customer group by Country")
df1.display()
```

```python
# windows function...
from pyspark.sql.window import Window
from pyspark.sql.functions import *

path=r'/Volumes/batch28_catalog/test_b28_schema/bronze/batch14_window.csv'
emp_window =Window.partitionBy('depart').orderBy('salary')
df=spark.read.csv(path,header=True,inferSchema=True)
df1=df.withColumn('row_num',row_number().over(emp_window))\
        .withColumn('rank',rank().over(emp_window))\
        .withColumn('dense_rank',dense_rank().over(emp_window))
        
        
df1.display()

```

```python
df2=df.withColumn('lead',lead('id').over(emp_window))\
        .withColumn('lag',lag('id').over(emp_window))
df2.display()
```
