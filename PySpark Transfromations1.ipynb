```python
# Datetime functions-----------
path=r'/Volumes/batch27_catalog/batch27/batch27_volume/bronze/batch18_student.csv'

df=spark.read.csv(path,header=True,inferSchema=True)
df.display()
```

```python
# add one column in dataframe with todays date..
# current_date - function in pyspark which gives us todays date
from pyspark.sql.functions import *
df1=df.withColumn("process_run_date",current_date())
display(df1)

from datetime import date
# using python function
curreent_date =date.today()
print(curreent_date)
df2=df.withColumn("process_run_date1",lit(curreent_date))
display(df2)
```

```python
# add new column with current time in dataframe
# current_timestamp:- this is method which add date_time value in column -(utc zone)
# from_utc_timestamp :- convert utc zone to require zone.
from pyspark.sql.functions import *
df3=df2.withColumn("process_run_time",current_timestamp())\
        .withColumn("process_run_time_ist",from_utc_timestamp(current_timestamp(),"Asia/Kolkata"))
display(df3) 
```

```python
# split date column to three columns -year,month,day

df4=df3.withColumn("process_year",year(current_date()))\
    .withColumn("process_month",month(current_date()))\
    .withColumn("process_day",dayofmonth(current_date()))\
    .withColumn('Day',dayofweek(current_date()))\
    .withColumn("process_hours",hour(col('process_run_time_ist')))\
    .withColumn("process_minutes",minute(col('process_run_time_ist')))\
    .withColumn("process_seconds",second(col('process_run_time_ist')))
df4.display()

 
```

```python
# findout datediff
path1=r'/Volumes/batch27_catalog/batch27/batch27_volume/bronze/customers-2000000.csv'
df10 = spark.read.csv(path1,header=True,inferSchema=True)
df10.display()


```

```python
# datediff :- this is method whoch provide date diff between two dates
df11=df10.withColumn('datediff',datediff(current_date(),col('Subscription Date')))\
        .withColumn("month_diff",months_between(current_date(),col('Subscription Date')))\
        .withColumn("year_diff",round(col('datediff')/365))\
        .withColumn("month_diff1",round((col('datediff')%365)/30))
       
df11.display()
```

```python
# orderby function:- by defualt order is Ascending order
df12=df11.orderBy(col("datediff")).select("Index","Customer Id","First Name","Last Name","datediff")
df12.display()
```

```python
df13=df12.orderBy(col("datediff").desc()).select("Index","Customer Id","First Name","Last Name","datediff").distinct()
df13.display()
```

```python
# remove duplicated from dataframe...
# distinct() :- this method checks all column values for record and if all values are saem then it delete duplicate record
# dropdupicates() :- this method checks only specific column value for record and if all values are same then it delete duplicate record
df13=df12.distinct()
df13.display()
```

```python
df14=df13.orderBy(col("datediff").desc()).select("Index","Customer Id","First Name","Last Name","datediff").dropDuplicates(["datediff"])
df14.display()
```

```python
# sort :-this mehod also used to arrange dataframe in sorted way
df15=df12.sort(col("datediff").desc()).select("Index","Customer Id","First Name","Last Name","datediff")
df15.display()
df16=df12.sort(col("datediff")).select("Index","Customer Id","First Name","Last Name","datediff")
df16.display()
```
