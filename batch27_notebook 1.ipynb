```python
from pyspark.sql.types import *
emp_data=[ 
            ("Dinesh",33,"BE","Sangli"),
            ("Shree",13,"8th","Karad"),
            ("Pari",6,"1st","Pune")
        ]

schema=StructType([
	StructField("name",StringType()),
	StructField("age",IntegerType()),
	StructField("Qualification",StringType()),
	StructField("City",StringType())
 ]
)

df=spark.createDataFrame(emp_data,schema)
df.show()
```

```python
# write dataframe 
#coalesce - reduce number of partitions
df.coalesce(1).write.mode("append").option('header','true').option("delimiter","|").csv("/Volumes/batch27_catalog/batch27/batch27_volume/result/")
```
