```python
# fill na 
data = [
    ("Alice", 25, "2026-07-05"),
    (None, 30, None),
    ("Bob", None, "2026-07-07"),
    (None, None, None)
]

columns = ["name", "age", "join_date"]

# Create DataFrame
df = spark.createDataFrame(data, columns)
df.display()
```

```python
df1 = df.fillna({"name": " "})
df1.display()
```

```python
df2 = df1.fillna({"age": 0})
df2.display()
```

```python
# convert datestring column to date type
df3 = df2.withColumn("join_date", to_date(df["join_date"], "yyyy-MM-dd"))

df3.display()
```

```python
# fillna() :- check if column value is null and fill with given value
df4=df3.fillna({"join_date": "2999-12-30"})
df4.display()
```

```python
# isNull() - function which check column value is NULL
df4 = df3.withColumn(
    "join_date",
    when(col("join_date").isNull(), to_date(lit("2999-12-30"), "yyyy-MM-dd"))
    .otherwise(col("join_date"))
)
df4.display()
```
