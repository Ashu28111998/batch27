```python
# read csv file ..............
path1 =r'/Volumes/batch27_catalog/batch27/batch27_volume/bronze/customers-2000000.csv'

df1=spark.read.format("csv").option("header","true").option("inferSchema","true").load(path1)
df1.display()
```

```python
# sorting out column names as per there types
string_col=[]
int_col=[]
other_col=[]
for col_details in df1.dtypes:
    # col_name=col_details[0]
    # col_type=col_details[1]
    if col_details[1] =='int':
        int_col.append(col_details[0])
    elif col_details[1] =='string':
        string_col.append(col_details[0])
    else:
        other_col.append(col_details[0])
print(f"columns from int list is:{int_col}")
print(f"columns from string list is:{string_col}")
print(f"columns from other list is:{other_col}")

```

```python
# create dataframes with specific columns with types
df2=df1.select(int_col)
df2.display()
```

```python
# select only specific columns from dataframe

df2=df1.select("Customer Id","First Name","Last Name")
df2.display()
```

```python
# select all columns from dataframe

df2=df1.select("*")
df2.display()
```

```python
# column rename in dataframe..
df3=df2.withColumnRenamed("Customer Id","customer_id")
df3.display()
```

```python
#rename all column names space with '_' and all in small case

for col_name in df3.columns:
    df3=df3.withColumnRenamed(col_name,col_name.replace(" ","_").lower())
df3.display()
```

```python
# add new index columns depends on index column
# col - it is function to select column
# drop - it is used to delete column from dataframe
from pyspark.sql.functions import col
df6 =df1.withColumn("new_index",col("Index") +1000)\
        .drop(col("Index"))\
        .withColumnRenamed("new_index","Index")
df6.display()
```

```python
# create new column with combine two string type column ( with concat function).......
# concat :- this us function is used to combine two string type column


from pyspark.sql.functions import concat,lit
df7=df6.withColumn("full_name",concat(col("First Name"),lit(" "),col("Last Name"))).select("full_name")
df7.display()
```

```python
#concat_ws :- this us function is used to combine two string type column with separator
from pyspark.sql.functions import concat_ws,lit
df7=df6.withColumn("full_name",concat_ws("_",col("First Name"),col("Last Name"))).select("full_name")
df7.display()
```

```python
# split :- method used to convert string type column into list
from pyspark.sql.functions import split
df8=df6.withColumn("user_name",split(col("email"),"@")[0])\
       .withColumn("domain_name",split(col("email"),"@")[1])
df8.display()
```
